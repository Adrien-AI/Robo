```python
import numpy as np
from sklearn import linear_model
rando=np.random.RandomState(0)
x=rando.randn(10, 5)
y=rando.randn(10)
print('\n x data:\n',x)
print('\n y data:\n',y)
```

```python
sgd_reg=linear_model.SGDRegressor(max_iter=1000,penalty="elasticnet",loss='huber',tol=1e-3,average=True)
sgd_reg.fit(x,y)
print('weight vector(s):',sgd_reg.coef_)
print('intercept:',sgd_reg.intercept_)
print('intercept:',sgd_reg.intercept_)
print('number of weights updated during training:',sgd_reg.t_)
```
