```python
import tensorflow as tf
```

```python
import numpy as np
```

```python
from tensorflow import keras
```

```python
import pandas as pd
```

```python
from sklearn.datasets import load_iris
```

```python
from sklearn.model_selection import train_test_split
```

```python
Iris_data=load_iris
```

```python
Iris_data
```

```python
Iris_data["target_names"]
```

```python
Iris_data["feature_names"]
```

```python
x=Iris_data['data']
```

```python
y=Iris_data['target']
```

```python
y.shape
```

```python
Data_Frame=pd.DataFrame(data=np.concatenate((x,y.reshape(-1,1)),axis=1),columns=['sepal length','sepal width','petal length','petal width',"target"])
```

```python
Data_Frame
```

```python
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.1, shuffle=True)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
input = tf.keras.Input(shape=(4, ))
X=tf.keras.layers.Dense(3, activation = tf.keras.activations.sigmoid)(input)
```

```python
model = tf.keras.models.Model(input, X)
```

```python
model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate = 0.01), loss = tf.keras.losses.sparse_categorical_crossentropy, metrics = ["accuracy"])
```

```python
train = model.fit(x_train, y_train.reshape(-1), validation_data=(x_test, y_test.reshape(-1)), epochs = 100)
```

```python
New_data=np.array([6.2,3.4,5.4,2])
```

```python
y_pred=model.predict(New_data.reshape(1,4))
```

```python
print("predicted:",y_pred)
print("predicted class:",np.argmax(y_pred))
```
