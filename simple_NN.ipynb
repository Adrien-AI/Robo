```python
import numpy as np
import tensorflow as tf
from tensorflow import keras

# Inputs (x) and Outputs (y)
# Relationship: y = 2x - 1
xs = np.array([-1.0, 0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)
ys = np.array([-3.0, -1.0, 1.0, 3.0, 5.0, 7.0], dtype=float)
```

```python
# Define a 1-layer neural network
model = keras.Sequential([
    keras.layers.Dense(units=1, input_shape=[1])
])

# Compile with an optimizer and loss function
model.compile(optimizer='sgd', loss='mean_squared_error')
```

```python
# Train the model
model.fit(xs, ys, epochs=500, verbose=0) # verbose=0 hides training logs
print("Training complete!")
```

```python
# Wrap the input in a numpy array to match the expected format
prediction = model.predict(np.array([[10.0]]))

# Extract the scalar value from the prediction array
print(f"Prediction for x=10: {prediction[0][0]:.4f}")
```
